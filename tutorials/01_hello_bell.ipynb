# Hello Bell — your first QVM circuit

Author in OpenQASM, simulate exactly, and ship the same circuit to Qiskit **and** Cirq without rewriting a line.


## 1. Ingest OpenQASM


```python
from qvm.parser import OpenQASM2Parser

QASM = """
OPENQASM 2.0;
include "qelib1.inc";
qreg q[2];
creg c[2];
h q[0];
cx q[0],q[1];
measure q[0] -> c[0];
measure q[1] -> c[1];
"""
bell = OpenQASM2Parser.parse(QASM)
print(bell)
```

```python
from qvm.simulator import Simulator

counts = Simulator().sample(bell, shots=2000, seed=1)
print({k: v for k, v in sorted(counts.items())})
assert set(counts) <= {"00", "11"}, "perfect correlations expected"
```

## 2. Ship it anywhere (converts-or-raises guarantee)


```python
qiskit_bell = bell.to_qiskit()
cirq_bell   = bell.to_cirq()
print(type(qiskit_bell).__module__, "->", qiskit_bell.count_ops())
print(type(cirq_bell).__module__, "->", len(list(cirq_bell.all_operations())), "ops")
```

## 3. Cross-validate against Aer


```python
counts = bell.run_qiskit_simulator(shots=1000)
print(counts)   # only '00' and '11' — perfect correlations
```
